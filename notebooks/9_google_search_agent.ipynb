```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_community.utilities import GoogleSerperAPIWrapper
from langchain_groq import ChatGroq
from langchain.agents import create_agent
from langgraph.checkpoint.memory import MemorySaver
```

```python
llm = ChatGroq(model = "openai/gpt-oss-120b")
search = GoogleSerperAPIWrapper()
```

```python
search.run ("which is the most famous ganesh temple in indore, madya pradesh, india?")
```

```python
agent = create_agent(
    model = llm,
    tools = [search.run],
    system_prompt = "You are a helpful assistant that can search the web for information.",
    checkpointer = MemorySaver()
)
```

```python
question = "which is the most famous ganesh temple in indore, madya pradesh, india?"
```

```python
response = agent.invoke(
    {"messages" : [{"role" : "user", "content" : "which is the most famous ganesh temple in indore, madya pradesh, india?"}]},
    {"configurable" : {"thread_id" : "anuradha"}}
    )
print(response["messages"][-1].content)
```
