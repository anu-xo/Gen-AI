```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_core.tools import tool
```

```python
@tool
def add_two_num(a : int, b : int) -> int:
    """Add two numbers together.
    Args:
        a: The first number.
        b: The second number.
    Returns:
        The sum of the two numbers.
    """
    return a + b

@tool
def mul_two_num(a : int, b : int) -> int:
    """Multiply two numbers together.
    Args:
        a: The first number.
        b: The second number.
    Returns:
        The product of the two numbers.
    """
    return a * b
```

```python
add_two_num.invoke({"a": 23, "b": 34})
mul_two_num.invoke({"a": 23, "b": 34})
```

```python
from langchain_groq import ChatGroq
from langchain.agents import create_agent
llm = ChatGroq(model = "openai/gpt-oss-120b")
```

```python
agent = create_agent(
    model = llm,
    tools = [add_two_num, mul_two_num],
    system_prompt = "You are a helpful assistant that can add and multiply numbers."
)
```

```python
response = agent.invoke({"messages" : [{"role" : "user", "content" : "What is the answer of the expression: 2*398 + 60?"}]})
print(response["messages"][-1].content)
```
