```python
from langchain_core.tools import tool
```

```python
@tool
def add_two_num(a : int, b : int) -> int:
    """Add two numbers together.
    Args:
        a: The first number.
        b: The second number.
    Returns:
        The sum of the two numbers.
    """
    return a + b

@tool
def mul_two_num(a : int, b : int) -> int:
    """Multiply two numbers together.
    Args:
        a: The first number.
        b: The second number.
    Returns:
        The product of the two numbers.
    """
    return a * b
```

```python
add_two_num.invoke({"a": 23, "b": 34})
```
