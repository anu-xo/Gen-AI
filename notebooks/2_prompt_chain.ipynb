####Prompt Template -dynamic prompts

```python
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
```

```python
llm = ChatOpenAI(model = "gpt-4o")
```

```python
prompts = ChatPromptTemplate.from_messages([
    {"role" : "system", "content" : "You are a {programming_language} developer."},
    {"role" : "user", "content" : "{query}"}
])
```

```python
final_prompts = prompts.format_messages(programming_language = "python", query = "Explain insertion in linked list.")
final_prompts
```

```python
res = llm.invoke(prompts.invoke({"programming_language": "python", "query": "Explain insertion in linked list."}).messages)
print(res.content)
```

```python
chains = prompts | llm
chains
```

```python
res = chains.invoke({"programming_language": "python", "query": "Explain insertion in linked list."})
print(res.content)
```

```python
res = llm.invoke(final_prompts)
print(res.content)
```

```python
res = llm.invoke(prompts)
print(res.content)
```
