```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_groq import ChatGroq
llm = ChatGroq( model = "openai/gpt-oss-120b")
```

```python
prompts = [
    {"role" : "user", "content" : "My name is Anuradha Joshi."},
    {"role" : "ai", "content" :'Nice to meet you, Anuradha Joshi! How can I assist you today?'},
    {"role" : "user", "content" : "What is my name?"}
]

res = llm.invoke(prompts)
res.content
```

```python
history = []

while True:
    query = input("User: ")
    history.append({"role" : "user", "content" : query})

    if query.lower() in ["exit", "quit", "bye", "goodbye"]:
        print("Okay, See you next time! GoodBye!")
        break
    
    print("User : ", query, "\n")

    res = llm.invoke(history)
    history.append({"role" : "ai", "content" : res.content})

    print("AI : ", res.content, "\n")
```
